**Пояснення:**
Kernel (ядро) — це фоновий процес Python, який виконує код і зберігає всі змінні в оперативній пам'яті під час поточної сесії. Команда "Restart Kernel" завершує цей процес і запускає його наново, повністю очищуючи пам'ять. Тому ядро "забуває" про всі раніше створені змінні. Щоб змінна `x` знову існувала, потрібно заново виконати осередок з її оголошенням.
